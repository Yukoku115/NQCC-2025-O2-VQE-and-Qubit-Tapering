```python
from qiskit import QuantumCircuit, transpile
from qiskit_alice_bob_provider import AliceBobLocalProvider
import numpy as np

provider = AliceBobLocalProvider()
backend = provider.get_backend("EMU:15Q:LOGICAL_EARLY")
```

```python
"""
Here, we construct a parameterised 2 qubit ansatz.
The circuit starts from |00>, goes through the first layer of single qubit rotations, to
make a superposition of |00>, |01>, |10> and |11>. Then, a CNOT gate is applied to entangle the two qubits.
Last layer adjusts the amplitudes and phases while keeping the entanglement.
"""
def make_2q_circuit(params):
    qc = QuantumCircuit(2)
    qc.ry(params[0], 0)
    qc.rz(params[1], 0)
    qc.ry(params[2], 1)
    qc.rz(params[3], 1)
    qc.cx(0, 1)
    qc.ry(params[4], 0)
    qc.rz(params[5], 0)
    qc.ry(params[6], 1)
    qc.rz(params[7], 1)
    return qc

test_params = np.linspace(0.1, 0.8, 8)
qc = make_2q_circuit(test_params)
print(qc.draw())

transpiled_qc = transpile(qc, backend=backend)
print(transpiled_qc.draw())
```

```python
gate_counts = transpiled_qc.count_ops()
for gate, count in gate_counts.items():
    print(f"{gate}: {count}")
```

```python
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

qc_run = qc.copy()
qc_run.measure_all()
transpiled_qc_run = transpile(qc_run, backend=backend)

job = backend.run(transpiled_qc_run, shots=1024)
result = job.result()
counts = result.get_counts()
print("Measurement counts:", counts)
plot_histogram(counts)
```

Ok so we just tested the 4 possible computational basis outcomes which are equally probable. Now, we will plug in the optimal angles found by VQE for the trigle GS $T_0$

```python
opt_params_T0 = np.array([4.72088452,  3.15168806, -0.02399731, -0.28146557,  
    4.26526778,  4.8850912,   4.26612921,  5.43348331])

qc_T0 = make_2q_circuit(opt_params_T0)
qc_T0.measure_all()
transpiled_qc_T0 = transpile(qc_T0, backend=backend)

job_T0 = backend.run(transpiled_qc_T0, shots=1024)
counts_T0 = job_T0.result().get_counts()
print("Measurement counts for T0:", counts_T0)
plot_histogram(counts_T0)
```

```python
# The 4 Hamiltonian coefficients
c_I  = -146.504130
c_ZZ =   -1.128285
c_YY =   -0.013482
c_XX =   -0.013482

def compute_parity(counts):
    """Computes <P_0 P_1> = (N_00 + N_11 - N_01 - N_10) / Total Shots"""
    total = sum(counts.values())
    even = counts.get('00', 0) + counts.get('11', 0)
    odd  = counts.get('01', 0) + counts.get('10', 0)
    return (even - odd) / total

# 1. Circuit for ZZ measurement (standard basis)
qc_zz = make_2q_circuit(opt_params_T0)
qc_zz.measure_all()
tqc_zz = transpile(qc_zz, backend=backend)
counts_zz = backend.run(tqc_zz, shots=2000).result().get_counts()
exp_zz = compute_parity(counts_zz)

# 2. Circuit for XX measurement (rotate with H before measurement)
qc_xx = make_2q_circuit(opt_params_T0)
qc_xx.h(0); qc_xx.h(1)
qc_xx.measure_all()
tqc_xx = transpile(qc_xx, backend=backend)
counts_xx = backend.run(tqc_xx, shots=2000).result().get_counts()
exp_xx = compute_parity(counts_xx)

# 3. Circuit for YY measurement (rotate with Sdg then H before measurement)
qc_yy = make_2q_circuit(opt_params_T0)
qc_yy.sdg(0); qc_yy.h(0)
qc_yy.sdg(1); qc_yy.h(1)
qc_yy.measure_all()
tqc_yy = transpile(qc_yy, backend=backend)
counts_yy = backend.run(tqc_yy, shots=2000).result().get_counts()
exp_yy = compute_parity(counts_yy)

# 4. Total Energy on Alice & Bob's Cat-Qubit Emulator
E_emulator = c_I + (c_ZZ * exp_zz) + (c_YY * exp_yy) + (c_XX * exp_xx)
E_exact = -147.632415

print("=" * 55)
print("  ALICE & BOB LOGICAL EMULATOR - ENERGY EVALUATION")
print("=" * 55)
print(f"  <Z0 Z1> = {exp_zz:+.4f}")
print(f"  <X0 X1> = {exp_xx:+.4f}")
print(f"  <Y0 Y1> = {exp_yy:+.4f}")
print("-" * 55)
print(f"  T0 Energy (Alice & Bob): {E_emulator:.6f} Ha")
print(f"  Exact CASCI Benchmark:   {E_exact:.6f} Ha")
print(f"  Error (Shot Variance):   {abs(E_emulator - E_exact) * 1000:.2f} mHa")
print("=" * 55)
```

### Notes on Emulator Results & Next Steps

1. **Why the emulator energy is −146.53 Ha instead of −147.632 Ha**:
   - In an ideal noiseless simulation, opt_params_T0 gives ~49% '00' and ~51% '11' (and 0% for '01' and '10'), which means $\langle Z_0 Z_1 \rangle = +1.0000$ and the energy is **−147.632390 Ha** (matching the exact benchmark −147.632415 Ha).
   - On EMU:15Q:LOGICAL_EARLY, logical qubits only support discrete gates (, S, T, CX$). Approximating the 8 continuous angles (, RZ$) turns the 9-gate circuit into **2,462 gates** (including **1,061 /T^\dagger$ gates**).
   - With the emulator's default noise settings (verage_nb_photons=7, distance=13, kappa_1/kappa_2 = 1e-3), gate errors (~^{-3}$ per gate) build up over 2,462 gates and randomize the state (~25% for each bitstring). This drives $\langle Z_0 Z_1 \rangle$ close to 0 and leaves only the identity term (−146.504 Ha).
